```python
import pandas as pd
```

```python
climate_data=pd.read_csv("data/climate_诚海诚源.csv")
climate_data["time"]=pd.to_datetime(climate_data["time"])
my_data=pd.read_excel("data/昭诚电科-城海诚源.xlsx",sheet_name='ec025-06-00')
my_data["Time"]=pd.to_datetime(my_data["Time"])
new_data=my_data[["Time","shortwave_radiation_ecmwf_ifs025"]].merge(climate_data,left_on="Time",right_on="time").drop(columns=["irradiance","time"]).rename(columns={"shortwave_radiation_ecmwf_ifs025":"irradiance"})
new_data.to_csv("城海诚源.csv",index=False)
```
